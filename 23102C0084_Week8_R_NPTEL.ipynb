## 1. Operations on Lists

### 1.1 Merging Lists (`c()`)

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

list1
list2
```

```r
list12 <- c(list1, list2)
list12
```

### 1.2 Lists to Vector (`unlist()`)

```r
unlist(list1)
unlist(list2)

mode(list1)
mode(unlist(list1))
```

### 1.3 Appending Lists (`append()`)

```r
append(list1, 100)
```

```r
append(list2, "coffee")
```

Appending **after** a specific position:

```r
append(list1, 100, after = 2)
```

```r
append(list2, "coffee", after = 2)
```

### 1.4 Removing From a List (`listname[-position]`)

```r
list1[-2]   # remove 2nd element
```

```r
list2[-1]   # remove 1st element
```

### 1.5 Extracting From a List (range of indexes)

```r
list1b <- list(1, 2, 3, 4, 5, 6)
list1b[2:4]
list1b[c(1, 3, 5)]
```

```r
list2b <- list("water", "juice", "lemonade", "tea", "coffee", "milk")
list2b[2:4]
list2b[c(1, 3, 5)]
```

## 2. Vector Indexing

### 2.1 Logical / Conditional Indexing

```r
x <- 1:10
x
x[(x > 5)]
x[(x %% 2 == 0)]   # even values
x[(x %% 2 == 1)]   # odd values
```

### 2.2 Vector with Missing Observations (`NA`)

```r
x[5] <- NA
x

y <- x[!is.na(x)]   # ! means negation
y

mean(x)   # NA present
mean(y)   # NA removed
```

### 2.3 Vector of Negative Integers

```r
x2 <- 1:10
x2[-(1:5)]
x2[(6:10)]   # equivalent result
```

### 2.4 Named Vectors / Lists

```r
z <- list(a1 = 1, a2 = "c", a3 = 1:3)
z
names(z)
```

```r
names(z)[3] <- "c2"   # rename only the 3rd element
z
```

```r
xv <- c(water = 1, juice = 2, lemonade = 3)
names(xv)
xv["juice"]
```

### 2.5 Empty Index

```r
x3 <- 1:10
x3[]   # empty index returns the full vector
```

## 3. Factors

### 3.1 Basic Factor Creation

```r
y_die <- c(1, 4, 3, 5, 4, 2, 4)
possible.dieface <- c(1, 2, 3, 4, 5, 6)
labels.dieface   <- c("one", "two", "three", "four", "five", "six")

facy <- factor(y_die, levels = possible.dieface, labels = labels.dieface)
facy
```

### 3.2 `as.factor()`

```r
xf <- c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)
yf <- as.factor(xf)
yf
```

```r
xf2 <- factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))
xf2   # levels ordered alphabetically
```

### 3.3 `class()`

```r
class(9)
class("9")
class(print)

mat <- matrix(nrow = 2, ncol = 2, data = 1:4)
class(mat)
```

### 3.4 `unclass()`

```r
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)
brands_fac
unclass(brands_fac)
```

```r
colours <- c("blue", "green", "red")
colours[unclass(brands_fac)]   # map factor codes back to colours
```

### 3.5 Custom Level Order

```r
xlvl <- factor(c("lemonade", "lemonade", "juice", "lemonade", "water"),
               levels = c("water", "juice", "lemonade"))
xlvl
unclass(xlvl)
levels(xlvl)
```

### 3.6 Ordered Factors

```r
income <- ordered(c("high", "high", "low", "medium", "medium"),
                   levels = c("low", "medium", "high"))
income
unclass(income)
```

## 4. Strings — Display & Formatting

### 4.1 `print()`

```r
print(sqrt(2))
print(sqrt(2), digits = 5)
print(sqrt(2), digits = 10)
```

```r
print("apple")
print(c("apple", "banana"))
print(c("apple", "banana", 6, 10))
```

### 4.2 `format()`: digits & nsmall

```r
format(0.5, digits = 10, nsmall = 15)
```

### 4.3 `format()`: width & justify

```r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "centre")
format(c("A", "BB", "CCC", "DDDD"), width = 14, justify = "centre")
```

```r
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "left")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "right")
format(c("A", "BB", "CCC", "DDDD"), width = 7, justify = "none")
```

### 4.4 Printing Matrices

```r
matx <- matrix(nrow = 3, ncol = 2, data = 1:6, byrow = TRUE)
print(matx)
```

### 4.5 `big.mark` for Large Numbers

```r
format(1234567, big.mark = ",")
format(12345678, big.mark = ",")
format(123456789, big.mark = ",")
format(123456789, big.mark = " ")
```
