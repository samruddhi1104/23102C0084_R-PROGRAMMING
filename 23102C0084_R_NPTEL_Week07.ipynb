## Lecture 26 — Sequences

A **sequence** is a set of related numbers/items that follow each other in a particular order.

### 1.1 Continuous sequences with constant unit increment using `:`

````r
1:10
10:1
5:15
15:5
````

````r
-1:-10
-10:-1
-5:-15
-15:-5
````

### 1.2 Sequences with decimal (fractional) start values

````r
1.23:10
1.23:10.54
10.54:2.23
````

````r
-1.23:-10
-5.23:6
````

### 1.3 The `seq()` function

```
seq(from = 1, to = 1, by = ((to - from)/(length.out - 1)), length.out = NULL, along.with = NULL, ...)
```

````r
help(seq)
````

````r
seq(10)
seq(1:10)   # same as seq(10)
````

### 1.4 Sequences with a predefined variable and constant increment

````r
x = 2
seq(1, x, x/10)

x = 50
seq(0, x, x/10)
````

### 1.5 Storing the outcome of sequences

````r
x = seq(1, 50, 1/2)
x

y = 2*x
y
````

### 1.6 Sequences: Index vector

Using `seq(along = x)` to build an index vector, then accessing elements through it.

````r
x = c(9,8,7,6)
ind = seq(along = x)
ind

# Accessing an element through the index vector
x[ ind[2] ]
````

### 1.7 Generating current time and date

````r
Sys.time()
Sys.Date()
````

## Lecture 27 — Sequences of Dates and Alphabets

### 2.1 `seq()` for dates
```
seq(from, to, by, length.out = NULL, along.with = NULL, ...)
```
- `from` : starting date (required)
- `to`   : end date (optional)
- `by`   : increment — `"day"`, `"week"`, `"month"`, `"quarter"`, or `"year"`
- `length.out` : desired length of the sequence


**Sequence of first day of each year**

````r
seq(as.Date("2010-01-01"), as.Date("2017-01-01"), by = "years")
````

**Sequence of days**

````r
seq(as.Date("2017-01-01"), by = "days", length = 6)
````

**Sequence of months**

````r
seq(as.Date("2017-01-01"), by = "months", length = 6)
````

**Sequence by years (using `length`)**

````r
seq(as.Date("2017-01-01"), by = "years", length = 6)
````

**Sequence with defined start and end dates, going backwards by month**

````r
startdate = as.Date("2016-1-1")
enddate   = as.Date("2017-1-1")
out = seq(enddate, startdate, by = "-1 month")
out
````

### 2.2 Generating sequences of alphabets

`letters` — sequence of lowercase alphabets

````r
letters
````

`letters[from_index:to_index]` — a specific range of lowercase alphabets

````r
letters[1:3]
letters[3:1]
letters[21:23]
letters[2]
````

`LETTERS` — sequence of uppercase alphabets

````r
LETTERS
````

`LETTERS[from_index:to_index]` — a specific range of uppercase alphabets

````r
LETTERS[1:3]
LETTERS[3:1]
LETTERS[21:23]
LETTERS[2]
````

## Lecture 28 — Repeats

The `rep()` function replicates the values of a vector a specified number of times.

```
rep(x)                # replicate values of x
rep(x, times = n)     # repeat x as a whole n times
rep(x, each = n)      # repeat each element n times
rep(x, length.out = n)  # or rep(x, length = n) / rep_len(x, length.out)
```


````r
help("rep")
````

### 3.1 Repeat an object n times

````r
rep(3.5, times = 10)
rep(1:4, 2)
````

### 3.2 Repeat an object n times vs repeat each cell n times

````r
x = 1:4
x

rep(x, times = 3)
rep(x, each = 3)
````

### 3.3 Combining `each` and `times` — order matters!

````r
rep(1:4, each = 2)
rep(1:4, each = 2, times = 3)
rep(1:4, times = 3, each = 2)
````

### 3.4 Repeating each element a different number of times

````r
rep(1:4, 2:5)
````

````r
ans = seq(from = 2, to = 8, by = 2)
ans

rep(1:4, ans)
````

### 3.5 Repeating a matrix

````r
x = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = T)
x

rep(x, 3)
````

### 3.6 Repetition of characters

````r
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)
````

### 3.7 Repetition of characters for a pre-specified length

````r
rep(2, length.out = 5)
rep(2, length = 5)
rep(c(2,3), length = 5)
rep(c(2,3,4), length = 5)
````

````r
rep("apple", length = 5)
rep(c("a", "b", "c"), length = 2)
rep(c("a", "b", "c"), length = 5)
````

## Lecture 29 — Sorting, Ordering and Mode

### 4.1 `sort()`
`sort()` arranges the values of a vector in ascending order (default) or descending order.

```
sort(x, decreasing = FALSE, ...)
sort(x, decreasing = FALSE, na.last = NA, ...)
```


````r
y = c(8,5,7,6)
y

sort(y)
sort(y, decreasing = TRUE)
````

### 4.2 `order()`
`order()` returns the *positions* that would arrange the vector in sorted order — i.e. it sorts according to the index.

```
order(x, decreasing = FALSE, ...)
order(x, decreasing = FALSE, na.last = TRUE, ...)
```


````r
y = c(9,8,5,7,6)
y

order(y)
order(y, decreasing = TRUE)
````

### 4.3 `mode()`
Every object has a **mode**, which indicates how it is stored in memory (number, character string, list, function, etc.).


````r
mode(2.432)
mode(c(3,4,5,6,7,8))
mode("India")
mode(c("India", "CANADA"))
````

````r
mode(factor(c("UP", "MP")))
mode(list("India", "USA"))
mode(data.frame(x = 1:2, y = c("India", "USA")))
mode(print)
````

## Lecture 30 — Lists

Unlike vectors, matrices, and arrays (which can hold only **one type of data**), a **list** can hold elements of **different types/modes**.

- Lists can be indexed by position: `x[[5]]`
- Lists can extract sublists: `x[c(2,5)]`
- List elements can have names: `x[["Students"]]` or `x$Students`


### 5.1 Arithmetic on matrices (for comparison)

````r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = T)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = T)
x1
x2

x1 + x2
````

### 5.2 Why lists are needed — mixing types breaks matrix arithmetic

````r
x1[2,1] = "hello"
x1

# This will now throw an error, since x1 has become a character matrix
tryCatch(
  x1 + x2,
  error = function(e) print(e)
)
````

### 5.3 Lists can contain matrices as objects

````r
x1 = matrix(nrow = 2, ncol = 2, data = 1:4, byrow = T)
x2 = matrix(nrow = 2, ncol = 2, data = 5:8, byrow = T)

matlist = list(x1, x2)
matlist
````

### 5.4 `[ ]` vs `[[ ]]` indexing on lists

````r
matlist[1]
matlist[2]
````

### 5.5 Lists containing different object types

````r
z1 = list(
  c("water", "juice", "lemonade"),
  rep(1:4, each = 2),
  matrix(data = 5:8, nrow = 2, ncol = 2, byrow = T)
)
z1
````

### 5.6 Accessing elements within a list — `[ ]` vs `[[ ]]`

````r
z1[[1]]        # the first list element (a character vector)

z1[1][2]       # WRONG: [1] gives a sub-list of length 1; [2] on that is out of range -> NULL

z1[[1]][2]     # CORRECT: [[1]] extracts the vector itself, [2] then indexes into it
````
