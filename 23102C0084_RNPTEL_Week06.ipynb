# Lecture 22 — Loops: the `for` loop

**Control structures in R:** control statements, functions, loops.  
Repetitive commands are executed by loops: **`for`**, **`while`**, **`repeat`**.

## The `for` loop
Use it when the **number of repetitions is known in advance** (e.g. run the same commands for cases *i = 1, 2, …, n*).

```
Start → initialise → condition? ──True──→ run instructions → update counter ─┐
                        │  ▲                                                  │
                      False └──────────────────────────────────────────────────┘
                        ↓
                       Stop
```

**Syntax**
```r
for (name in vector) { commands to be executed }
```
The variable `name` is set, in turn, to every value in `vector`, and the commands run once per value.

### Example 1 — squares of 1 to 5

````r
for (i in 1:5) { print(i^2) }
````

### Example 2 — the vector need not be a regular sequence

````r
for (i in c(2, 4, 6, 7)) { print(i^2) }
````

### Example 3 — loop inside a function: count elements with `x/2 > 3`

````r
x = c(2, 4, 6, 8, 10, 12)

excount = function(x) {
  count = 0
  for (xval in x) {
    if (xval/2 > 3)
      count = count + 1
  }
  print(count)
}

excount(x)      # 8, 10 and 12 qualify
````

## Nested `for` loops
A loop can sit **inside another loop**. The inner loop runs completely for every step of the outer loop.

### Example 4 — every child gets every sweet

````r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
````

## `break` — stop the loop early
`break` ends the loop before all items have been processed. Here the loop stops when it reaches `"tea"`.

````r
drink = c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
````

## `next` — skip one iteration
`next` skips the rest of the current iteration and continues with the next value, **without ending** the loop.

Skip `"lemonade"`:

````r
for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
````

Skip `"tea"`:

````r
for (x in drink) {
  if (x == "tea") {
    next
  }
  print(x)
}
````

**`break` vs `next`** — same condition, different effect:

````r
cat("with break:", "\n")
for (x in drink) { if (x == "tea") break;  print(x) }

cat("with next:", "\n")
for (x in drink) { if (x == "tea") next;   print(x) }
````

# Lecture 23 — Loops: `while` and `repeat`

## The `while` loop
Use it when the number of iterations is **not known beforehand** (e.g. an iterative algorithm that runs until it converges).

**Syntax**
```r
while (condition) { commands to be executed as long as condition is TRUE }
```
If the condition is **not true before entering** the loop, the body never runs.

### Example 1 — odd squares below 10

````r
i = 1
while (i < 10) {
  print(i^2)
  i = i + 2
}
````

### Example 2 — sum of the numbers from a starting value up to 25

The slide asks the user to type a number with `readline()`. `readline()` only works in an **interactive console**, so this version takes the number as an argument and uses `readline()` only if none is supplied.

> Note: the variable is called `sum`, as in the slide. This works, but it hides the built-in `sum()` inside the function — a different name such as `total` is safer in your own code.

````r
sumfunction = function(number = NA) {
  if (is.na(number)) {
    number = as.integer(readline(prompt = "Please select any number less than 25:   "))
  }
  if (is.na(number)) {
    cat("No number received. Call sumfunction(22) instead.\n")
    return(invisible(NULL))
  }
  sum = 0
  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While Loop:  ", sum))
}

sumfunction(22)
````

````r
22 + 23 + 24 + 25     # check
````

## The `repeat` loop
`repeat` tests **no condition** — not before the loop and not during it. You are responsible for ending it with **`break`**; forget it and the loop runs forever.

**Syntax**
```r
repeat { commands to be executed }
```

### Example 3 — same odd squares, ended with `break`

````r
i = 1
repeat {
  print(i^2)
  i = i + 2
  if (i > 10)
    break
}
````

### Example 4 — `next` inside `repeat`
`next` jumps back to the first command of the loop. Here it skips the printing until `i` reaches 10, then prints 100, 121, 144, 169 and breaks at 13.

````r
i = 1
repeat {
  i = i + 1
  if (i < 10) next
  print(i^2)
  if (i >= 13) break
}
````

### Which loop should I use?
| Loop | Use when | Stops when |
|---|---|---|
| `for` | number of repetitions known / iterate over a vector | vector exhausted |
| `while` | number of repetitions unknown | condition becomes `FALSE` (checked **before** each pass) |
| `repeat` | you want to control exit yourself | you call `break` |

# Lecture 24 — Functions

A function is a bunch of commands grouped into a sensible unit. It takes **input arguments**, does calculations (or graphics, or calls other functions) and **returns a result**. The result can be a complex object such as a list.

## Components
| Component | Meaning |
|---|---|
| **Function name** | stored as an object with this name |
| **Arguments** | the input values; a function may have none |
| **Function body** | statements that define what the function does |
| **Return value** | the value of the **last expression** evaluated in the body |

## Built-in functions
`sum()`, `prod()`, `mean()`, `max()`, … are already provided by R.

````r
v = c(3, 1, 4, 1, 5)
sum(v); prod(v); mean(v); max(v)
````

## Syntax
```r
Name <- function(Argument1, Argument2, ...) {
  expression(s)
}
```
`=` works in place of `<-`:
```r
Name = function(Argument1, Argument2, ...) { expression(s) }
```

### One argument

````r
abc = function(x) {
  x^2
}

abc(3)
abc(6)
abc(9)
````

### Two arguments

````r
abc = function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
````

### Another example — `sin²x + cos²x = 1`, so the function returns `x + 1`

````r
abc = function(x) {
  sin(x)^2 + cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)
````

### A function with no arguments

````r
abc = function() {
  for (i in 1:3) {
    print(i^3)
  }
}

abc()
````

## Arguments: names, default values and `...`
Arguments can have **meaningful names**, **default values**, and the special argument **`...`** (extra arguments passed on to other functions).

````r
# default values + named arguments
power = function(x, p = 2) {
  x^p
}

power(4)            # uses p = 2
power(4, p = 3)     # override the default
power(p = 3, x = 2) # order does not matter when names are used
````

````r
# the special argument `...`
describe = function(x, ...) {
  cat("Mean of x:", mean(x, ...), "\n")
}

describe(c(1, 2, NA, 4))                 # NA propagates
describe(c(1, 2, NA, 4), na.rm = TRUE)   # `...` passes na.rm on to mean()
````

### Return value: the last expression, or an explicit `return()`

````r
f1 = function(x) { x * 2 }            # returns last expression
f2 = function(x) { return(x * 2) }    # explicit return

f1(5); f2(5)

# returning a list bundles several results
stats = function(x) list(mean = mean(x), sd = sd(x))
stats(c(2, 4, 6, 8))
````

# Lecture 25 — Sequences

A **sequence** is a set of related numbers that follow each other in a particular order. R generates regular sequences with `seq()`.

```r
seq(from = 1, to = 1,
    by = ((to - from)/(length.out - 1)),
    length.out = NULL, along.with = NULL, ...)
```
Help page: `help("seq")` (or `?seq`).

````r
# help("seq")     # uncomment to open the documentation
````

### Default increment is +1 or −1

````r
seq(from = 2, to = 4)
seq(from = 4, to = 2)
seq(from = -4, to = 4)
````

### Constant increment / decrement with `by`

````r
seq(from = 10, to = 20, by = 2)    # increment of 2
seq(from = 20, to = 10, by = -2)   # decrement of 2
````

### Decreasing by a fraction: 3 down to −2 in steps of −0.5

````r
seq(from = 3, to = -2, by = -0.5)
````

### Predefined length with `length`
`length` is short for `length.out` (R allows partial argument names). The default step is +1.

````r
seq(to = 10, length = 10)
seq(from = 10, length = 10)
````

### Predefined length with a fractional increment / decrement

````r
seq(from = 10, length = 10, by = 0.1)
seq(from = 10, length = 10, by = -2)
seq(from = 10, length = 5,  by = -.2)
````

### Extras from the syntax line
`length.out` alone spaces values evenly between `from` and `to`; `along.with` copies the length of another object.

````r
seq(from = 0, to = 1, length.out = 5)
seq(along.with = c("a", "b", "c", "d"))    # 1 2 3 4
````

# Quick Reference

| Idea | Code |
|---|---|
| `for` loop | `for (i in 1:5) { print(i^2) }` |
| Nested loop | `for (x in a) { for (y in b) { print(paste(x, y)) } }` |
| Stop a loop | `break` |
| Skip an iteration | `next` |
| `while` loop | `while (i < 10) { ...; i = i + 2 }` |
| `repeat` loop | `repeat { ...; if (cond) break }` |
| Define a function | `f = function(x, y = 2) { x^y }` |
| Call it | `f(3)`, `f(3, y = 4)` |
| Sequence | `seq(from, to, by)` or `seq(from, by, length = n)` |

**Key points**
1. `for` = known number of passes; `while` = test *before* each pass; `repeat` = you must `break`.
2. `break` leaves the loop, `next` skips to the next iteration.
3. A function returns the value of its **last expression**.
4. `seq()` defaults to steps of ±1; use `by` for other steps and `length` for a fixed count.
