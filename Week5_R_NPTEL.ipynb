```r
# ============================================================
# NPTEL R PROGRAMMING - WEEK 5
# FOUNDATIONS OF R SOFTWARE
# Missing Data + Conditional Execution
# ============================================================

cat("============================================================\n")
cat("       NPTEL R PROGRAMMING - WEEK 5 IMPLEMENTATION\n")
cat("============================================================\n\n")


# ============================================================
# PART 1: MISSING DATA
# ============================================================

cat("\n================ PART 1: MISSING DATA ================\n")

# ------------------------------------------------------------
# 1. NA - Missing Value
# ------------------------------------------------------------

cat("\n--- 1. NA - Missing Value ---\n")

x <- NA

print(x)

# Check whether x is missing
print(is.na(x))


# ------------------------------------------------------------
# 2. is.na() with a vector
# ------------------------------------------------------------

cat("\n--- 2. is.na() with Vector ---\n")

x <- c(11, NA, 13, NA)

cat("Vector:\n")
print(x)

cat("Missing value locations:\n")
print(is.na(x))


# ------------------------------------------------------------
# 3. Mean with NA
# ------------------------------------------------------------

cat("\n--- 3. Mean with Missing Values ---\n")

x <- c(11, NA, 13, NA)

cat("Original vector:\n")
print(x)

cat("Mean without removing NA:\n")
print(mean(x))

cat("Mean after removing NA:\n")
print(mean(x, na.rm = TRUE))


# ------------------------------------------------------------
# 4. NA versus NULL
# ------------------------------------------------------------

cat("\n--- 4. NA versus NULL ---\n")

x <- NA
y <- NULL

cat("x = NA\n")
print(x)

cat("y = NULL\n")
print(y)

cat("Is x NA?\n")
print(is.na(x))

cat("Is y NULL?\n")
print(is.null(y))


# ------------------------------------------------------------
# 5. Find location of missing values using which()
# ------------------------------------------------------------

cat("\n--- 5. Location of Missing Values ---\n")

x <- c(11, NA, 13, NA)

cat("Vector:\n")
print(x)

cat("Positions containing NA:\n")
print(which(is.na(x)))


# ------------------------------------------------------------
# 6. Count number of missing values
# ------------------------------------------------------------

cat("\n--- 6. Count Missing Values ---\n")

x <- c(11, NA, 13, NA)

cat("Number of missing values:\n")
print(sum(is.na(x)))


# ------------------------------------------------------------
# 7. complete.cases()
# ------------------------------------------------------------

cat("\n--- 7. Complete Cases ---\n")

x <- c(11, NA, 13, NA)

cat("Vector:\n")
print(x)

cat("Complete cases:\n")
print(complete.cases(x))

cat("Positions of complete cases:\n")
print(which(complete.cases(x)))


# ------------------------------------------------------------
# 8. na.omit()
# ------------------------------------------------------------

cat("\n--- 8. Remove Missing Values using na.omit() ---\n")

x <- c(11, NA, 13, NA)

cat("Original vector:\n")
print(x)

y <- na.omit(x)

cat("Vector after removing NA:\n")
print(y)

cat("Mean of original vector:\n")
print(mean(x))

cat("Mean after removing NA:\n")
print(mean(y))


# ============================================================
# PART 2: IF STATEMENT
# ============================================================

cat("\n================ PART 2: IF STATEMENT ================\n")


# ------------------------------------------------------------
# Example 1: Multiply x by 3 if x > 4
# ------------------------------------------------------------

cat("\n--- Example 1: if statement ---\n")

x <- 5

if (x > 4) {
  print(x * 3)
}

x <- 3

if (x > 4) {
  print(x * 3)
}


# ------------------------------------------------------------
# Example 2: Print message if value > 3
# ------------------------------------------------------------

cat("\n--- Example 2: Check whether value is greater than 3 ---\n")

x <- 6

if (x > 3) {
  print("The value is more than 3")
}


# ------------------------------------------------------------
# Example 3: Condition is FALSE
# ------------------------------------------------------------

cat("\n--- Example 3: FALSE condition ---\n")

x <- 2

if (x > 3) {
  print("The value is more than 3")
}


# ============================================================
# PART 3: IF - ELSE
# ============================================================

cat("\n================ PART 3: IF - ELSE ================\n")


# ------------------------------------------------------------
# Example 4: x == 3
# ------------------------------------------------------------

cat("\n--- Example 4 ---\n")

x <- 5

if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}

cat("Value of x:\n")
print(x)


# ------------------------------------------------------------
# Example 5: x == 3
# ------------------------------------------------------------

cat("\n--- Example 5 ---\n")

x <- 3

if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}

cat("Value of x:\n")
print(x)


# ------------------------------------------------------------
# Example 6: Greater than or less than 3
# ------------------------------------------------------------

cat("\n--- Example 6 ---\n")

x <- 6

if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}


# ------------------------------------------------------------
# Example 7: Less than 3
# ------------------------------------------------------------

cat("\n--- Example 7 ---\n")

x <- 2

if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}


# ============================================================
# PART 4: NESTED IF - ELSE IF - ELSE
# ============================================================

cat("\n================ PART 4: NESTED IF-ELSE-IF ================\n")


# ------------------------------------------------------------
# Example 1: x = 5
# ------------------------------------------------------------

cat("\n--- Nested Example 1: x = 5 ---\n")

x <- 5

if (x == 3) {

  x <- x - 1

} else if (x < 3) {

  x <- x + 5

} else {

  x <- 2 * x

}

cat("Final value of x:\n")
print(x)


# ------------------------------------------------------------
# Example 2: x = 2
# ------------------------------------------------------------

cat("\n--- Nested Example 2: x = 2 ---\n")

x <- 2

if (x == 3) {

  x <- x - 1

} else if (x < 3) {

  x <- x + 5

} else {

  x <- 2 * x

}

cat("Final value of x:\n")
print(x)


# ------------------------------------------------------------
# Example 3: x = 3
# ------------------------------------------------------------

cat("\n--- Nested Example 3: x = 3 ---\n")

x <- 3

if (x == 3) {

  x <- x - 1

} else if (x < 3) {

  x <- x + 5

} else {

  x <- 2 * x

}

cat("Final value of x:\n")
print(x)


# ============================================================
# PART 5: IFELSE()
# ============================================================

cat("\n================ PART 5: IFELSE() ================\n")


# ------------------------------------------------------------
# Example 4: Square values less than 6
# ------------------------------------------------------------

cat("\n--- Example 4: ifelse() ---\n")

x <- 1:10

cat("Original vector:\n")
print(x)

result <- ifelse(x < 6, x^2, x + 1)

cat("Result:\n")
print(result)


# ------------------------------------------------------------
# Example 5: Even/Odd numbers
# ------------------------------------------------------------

cat("\n--- Example 5: Even or Odd ---\n")

x <- c(7, 9, 8, 4)

cat("Original vector:\n")
print(x)

result <- ifelse(
  x %% 2 == 0,
  "even number",
  "odd number"
)

cat("Result:\n")
print(result)


# ============================================================
# PART 6: SWITCH()
# ============================================================

cat("\n================ PART 6: SWITCH() ================\n")


# ------------------------------------------------------------
# Example 1: switch() with integer
# ------------------------------------------------------------

cat("\n--- switch() with integer ---\n")

result1 <- switch(
  2,
  "apple",
  "banana",
  "orange"
)

cat("switch(2, apple, banana, orange):\n")
print(result1)


result2 <- switch(
  1,
  "apple",
  "banana",
  "orange"
)

cat("switch(1, apple, banana, orange):\n")
print(result2)


# ------------------------------------------------------------
# Example 2: switch() with character string
# ------------------------------------------------------------

cat("\n--- switch() with character string ---\n")

result3 <- switch(
  "colour",
  colour = "blue",
  gender = "male",
  volume = 50
)

cat("Result for colour:\n")
print(result3)


result4 <- switch(
  "volume",
  colour = "blue",
  gender = "male",
  volume = 50
)

cat("Result for volume:\n")
print(result4)


# ------------------------------------------------------------
# Example 3: switch() with no match
# ------------------------------------------------------------

cat("\n--- switch() with no matching case ---\n")

result5 <- switch(
  4,
  "apple",
  "banana",
  "orange"
)

cat("Result for integer 4:\n")
print(result5)


result6 <- switch(
  "size",
  colour = "blue",
  gender = "male",
  volume = 50
)

cat("Result for size:\n")
print(result6)


# ============================================================
# PART 7: WHICH()
# ============================================================

cat("\n================ PART 7: WHICH() ================\n")


# ------------------------------------------------------------
# Example 4: which() with vector
# ------------------------------------------------------------

cat("\n--- which() with vector ---\n")

x <- c(10, 15, 8, 14, 6, 12)

cat("Vector:\n")
print(x)


cat("\nPosition where x == 14:\n")
print(which(x == 14))


cat("\nPositions where x != 12:\n")
print(which(x != 12))


cat("\nPositions where x > 10:\n")
print(which(x > 10))


# ============================================================
# PART 8: WHICH.MIN() AND WHICH.MAX()
# ============================================================

cat("\n================ PART 8: WHICH.MIN/MAX ================\n")


# ------------------------------------------------------------
# Matrix example
# ------------------------------------------------------------

cat("\n--- Matrix Example ---\n")

x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

cat("Matrix:\n")
print(x)


cat("\nPosition of minimum value:\n")
print(which.min(x))


cat("\nPosition of maximum value:\n")
print(which.max(x))


# ============================================================
# PART 9: WHICH() WITH MODULO OPERATOR
# ============================================================

cat("\n================ PART 9: WHICH() + MODULO ================\n")


cat("\n--- Find odd values in matrix ---\n")

x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

cat("Matrix:\n")
print(x)


# Find positions of odd values
odd_positions <- which(x %% 2 == 1)

cat("\nLinear positions of odd values:\n")
print(odd_positions)


# Find row and column positions
odd_indices <- which(
  x %% 2 == 1,
  arr.ind = TRUE
)

cat("\nRow and column positions of odd values:\n")
print(odd_indices)


# ============================================================
# PART 10: ADDITIONAL COMBINED EXAMPLES
# ============================================================

cat("\n================ PART 10: COMBINED EXAMPLES ================\n")


# ------------------------------------------------------------
# Example: Missing values + ifelse()
# ------------------------------------------------------------

cat("\n--- Missing Values + ifelse() ---\n")

marks <- c(85, 72, NA, 91, 65, NA, 78)

cat("Marks:\n")
print(marks)

cat("\nMissing value positions:\n")
print(which(is.na(marks)))

cat("\nNumber of missing marks:\n")
print(sum(is.na(marks)))

# Replace missing values with mean
mean_marks <- mean(marks, na.rm = TRUE)

marks_filled <- ifelse(
  is.na(marks),
  mean_marks,
  marks
)

cat("\nMean excluding NA:\n")
print(mean_marks)

cat("\nMarks after replacing NA with mean:\n")
print(marks_filled)


# ------------------------------------------------------------
# Example: Grade using if-else-if
# ------------------------------------------------------------

cat("\n--- Grade Calculation using Nested if-else-if ---\n")

marks <- 85

if (marks >= 90) {

  grade <- "A+"

} else if (marks >= 80) {

  grade <- "A"

} else if (marks >= 70) {

  grade <- "B"

} else if (marks >= 60) {

  grade <- "C"

} else {

  grade <- "F"

}

cat("Marks:", marks, "\n")
cat("Grade:", grade, "\n")


# ------------------------------------------------------------
# Example: Even/Odd using ifelse()
# ------------------------------------------------------------

cat("\n--- Even/Odd for Multiple Numbers ---\n")

numbers <- 1:20

even_odd <- ifelse(
  numbers %% 2 == 0,
  "Even",
  "Odd"
)

result_table <- data.frame(
  Number = numbers,
  Type = even_odd
)

print(result_table)


# ------------------------------------------------------------
# Example: Find values greater than 10
# ------------------------------------------------------------

cat("\n--- Find Values Greater Than 10 ---\n")

values <- c(5, 12, 8, 15, 20, 3, 17)

positions <- which(values > 10)

cat("Values:\n")
print(values)

cat("Positions where value > 10:\n")
print(positions)

cat("Values greater than 10:\n")
print(values[positions])


# ============================================================
# PART 11: VISUAL PRESENTATION
# ============================================================

cat("\n================ PART 11: VISUALIZATION ================\n")


# ------------------------------------------------------------
# Visualization 1: Missing values
# ------------------------------------------------------------

marks <- c(85, 72, NA, 91, 65, NA, 78, 88, NA, 95)

missing_status <- ifelse(
  is.na(marks),
  1,
  0
)

barplot(
  missing_status,
  names.arg = 1:length(marks),
  main = "Missing Value Locations",
  xlab = "Observation",
  ylab = "Missing (1 = Yes, 0 = No)",
  ylim = c(0, 1.2)
)


# ------------------------------------------------------------
# Visualization 2: Even/Odd
# ------------------------------------------------------------

numbers <- 1:20

barplot(
  numbers,
  main = "Numbers from 1 to 20",
  xlab = "Position",
  ylab = "Value"
)


# ------------------------------------------------------------
# Visualization 3: Values and threshold
# ------------------------------------------------------------

values <- c(5, 12, 8, 15, 20, 3, 17)

barplot(
  values,
  main = "Values with Threshold 10",
  xlab = "Observation",
  ylab = "Value"
)

abline(
  h = 10,
  lty = 2,
  lwd = 2
)


# ============================================================
# PART 12: FINAL SUMMARY OUTPUT
# ============================================================

cat("\n============================================================\n")
cat("                    WEEK 5 COMPLETED\n")
cat("============================================================\n")

cat("\nTopics implemented:\n")

topics <- c(
  "1. Missing Data - NA",
  "2. is.na()",
  "3. mean() with na.rm",
  "4. NA versus NULL",
  "5. which(is.na())",
  "6. sum(is.na())",
  "7. complete.cases()",
  "8. na.omit()",
  "9. if statement",
  "10. if-else statement",
  "11. Nested if-else-if",
  "12. ifelse()",
  "13. switch()",
  "14. which()",
  "15. which.min()",
  "16. which.max()",
  "17. which() with arr.ind=TRUE",
  "18. Modulo operator %%",
  "19. Combined examples",
  "20. Visualizations"
)

print(topics)

cat("\n============================================================\n")
cat("All Week 5 concepts have been implemented successfully.\n")
cat("============================================================\n")
```
